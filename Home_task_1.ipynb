```python
import numpy as np

arr = np.random.rand(5, 5)
arr = arr - arr.mean(axis=1, keepdims=True)
print(arr)
print(arr.mean(axis=1))
```

```python
import numpy as np

arr = np.random.rand(10, 3)                        # случайная матрица 10x3
idx = np.abs(arr - 0.5).argmin(axis=1)             # индексы ближайших к 0.5 в каждой строке
result = arr[np.arange(10), idx]                   # берём эти значения

print(arr)
print("Ближайшие к 0.5:", result)
```

```python
import numpy as np

arr = np.random.rand(6, 6)
row_sums = arr.sum(axis=1)
col_maxes = arr.max(axis=0)
result = row_sums / col_maxes

print(arr)
print(result)
```

```python
import numpy as np

a = np.array([[1, 4.9, 2.4], [7.1, 7, 9.8], [2, 4.6, 1.2]])
b = np.array([3, 5, 10, 0])

result = b[np.abs(a[:, :, None] - b).argmin(axis=-1)]
print(result)
```

```python
import numpy as np

a = np.array([1, 4.9, 2.4, 7.1, 7, 9.8, 2, -5, 4.6, 1.2])
result = np.convolve(a, np.ones(3) / 3, mode='valid')   # свёртка с ядром [1/3,1/3,1/3] = скользящее среднее, mode='valid' убирает неполные окна
print(result)
```

```python
import numpy as np

N, M = np.random.randint(2, 11, 2)
a = np.random.rand(N, M)
result = (a[:-1, :-1] + a[:-1, 1:] + a[1:, :-1] + a[1:, 1:]) / 4    # среднее по каждому окну 2x2
print(a)
print(result)
```
